```python
import numpy as np
```

```python
# Rows = cities, Columns = years (2018, 2019, 2020, 2021)
population = np.array([
    [500000, 520000, 540000, 560000],  # City 1
    [1200000, 1250000, 1300000, 1350000],  # City 2
    [800000, 790000, 780000, 770000],  # City 3 (declining)
    [300000, 320000, 340000, 360000],  # City 4
])
```

```python
cities = np.array(["City1", "City2", "City3", "City4"])
years = np.array([2018, 2019, 2020, 2021])
```

**Total Population per Year**

```python
total_per_year = np.sum(population, axis=0)
print("Total Population per Year:", total_per_year)
```

**Growth Rate per Year**

$$growth = \frac{pop_{year+1} - pop_{year}}{pop_{year}} \times 100$$

```python
growth_rate = np.diff(population, axis=1) / population[:, :-1] * 100
print("Yearly Growth Rate (%) per City:\n", growth_rate.round(2))
```

**City with Highest / Lowest Population (Latest Year)**

```python
latest_year_population = population[:, -1]
max_pop_city = cities[np.argmax(latest_year_population)]
min_pop_city = cities[np.argmin(latest_year_population)]
```

```python
print(f"City with Highest Population (2021): {max_pop_city}")
print(f"City with Lowest Population (2021): {min_pop_city}")
```

**Detect Declining Cities**

```python
declining_mask = np.all(np.diff(population, axis=1) <= 0, axis=1)
declining_cities = cities[declining_mask]
```

```python
print("Cities with Declining Population:", declining_cities)
```

**Average Population per City**

```python
avg_population = np.mean(population, axis=1)
for city, avg in zip(cities, avg_population):
    print(f"{city} Average Population: {avg:.0f}")
```
