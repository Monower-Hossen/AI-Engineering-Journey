```python
import numpy as np
```

```python
np.random.seed(42)  # reproducible results
```

```python
# Parameters
num_games = 1000
dice_sides = 6
```

**Roll Dice (Vectorized)**

```python
# Player 1 rolls: shape (num_games, 2)
player1_rolls = np.random.randint(1, dice_sides + 1, size=(num_games, 2))
player2_rolls = np.random.randint(1, dice_sides + 1, size=(num_games, 2))
```

```python
# Total per game
player1_total = np.sum(player1_rolls, axis=1)
player2_total = np.sum(player2_rolls, axis=1)
```

**Determine Winners**

```python
player1_wins = player1_total > player2_total
player2_wins = player2_total > player1_total
draws = player1_total == player2_total
```

```python
# Count results
p1_win_count = np.sum(player1_wins)
p2_win_count = np.sum(player2_wins)
draw_count = np.sum(draws)
```

```python
print(f"Player 1 Wins: {p1_win_count}")
print(f"Player 2 Wins: {p2_win_count}")
print(f"Draws: {draw_count}")
```

**Win Probabilities**

```python
p1 = p1_win_count / num_games
p2 = p2_win_count / num_games
dp = draw_count / num_games
```

```python
print(f"Player 1 Win Probability: {p1}")
print(f"Player 2 Win Probability: {p2}")
print(f"Draw Probability: {dp}")
```

**Advanced Stats**

```python
print(f"Player 1 Average Total Roll: {np.mean(player1_total)}")
print(f"Player 2 Average Total Roll: {np.mean(player2_total)}")
```

```python
HR_max1 = np.max(player1_total)
HR_max2 = np.max(player2_total)
```

```python
LR_min1 = np.min(player1_total)
LR_min2 =  np.min(player2_total)
```

```python
print(f"Highest Roll: {max(HR_max1 , HR_max2)}")
print(f"Lowest Roll: {min(LR_min1,LR_min2)}")
```
