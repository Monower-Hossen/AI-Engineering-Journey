```python
import numpy as np
```

```python
# rows = students, columns = subjects
marks = np.array([
    [78, 82, 91, 67],
    [55, 61, 58, 60],
    [92, 88, 95, 90],
    [40, 45, 50, 42],
    [85, 79, 88, 91],
    [66, 70, 72, 68]
])
```

```python
subjects = np.array(["Math", "Physics", "CS", "English"])
```

**Student Performance Metrics**

```python
avg = np.mean(marks, axis=1)
avg
```

```python
std = np.std(marks, axis=1)
std
```

```python
max_mark = np.max(marks, axis=1)
max_mark
```

```python
min_mark = np.min(marks, axis=1)
min_mark
```

**Student Ranking**

```python
rank = np.argsort(-avg) + 1
rank 
```

**Pass / Fail Logic (Real Rule)**

```python
pass_mask = np.all(marks >= 60, axis=1)
pass_mask
```

**Grade System**

```python
grades = np.where(
    avg >= 85, "A+",
    np.where(avg >= 75, "A",
    np.where(avg >= 60, "B", "F"))
)
grades
```

**Scholarship Eligibility**

```python
scholarship = (avg >= 80) & pass_mask
scholarship
```

**Subject Difficulty Analysis**

```python
subject_avg = np.mean(marks, axis=0)
subject_avg
```

```python
hardest_subject = subjects[np.argmin(subject_avg)]
hardest_subject
```

```python
easiest_subject = subjects[np.argmax(subject_avg)]
easiest_subject
```

**Identify At-Risk Students**

```python
at_risk_students = np.where(avg < 60)[0]
at_risk_students
```

**Final Report**

```python
report = np.column_stack((
    avg.round(2),
    grades,
    pass_mask,
    scholarship
))

print("Avg | Grade | Pass | Scholarship")
print(report)
```
