```python
import numpy as np
```

```python
# Categories: Food, Transport, Entertainment, Utilities
categories = np.array(["Food", "Transport", "Entertainment", "Utilities"])
```

```python
days = np.arange(1, 11)  # Track 10 days
days
```

```python
# Random daily expenses (rows = days, columns = categories)
np.random.seed(42)
expenses = np.random.randint(50, 500, size=(10, 4))
```

```python
print("Expenses (per day per category):\n", expenses)
```

**Total & Average Expense per Category**

```python
total_per_category = np.sum(expenses, axis=0)
```

```python
avg_per_category = np.mean(expenses, axis=0)
```

```python
for cat, total, avg in zip(categories, total_per_category, avg_per_category):
    print(f"{cat}: Total = {total}, Average = {avg:.2f}")
```

**Highest & Lowest Expense**

```python
max_expense = np.max(expenses)
min_expense = np.min(expenses)
```

```python
max_pos = np.unravel_index(np.argmax(expenses), expenses.shape)
min_pos = np.unravel_index(np.argmin(expenses), expenses.shape)
```

```python
print(f"Highest Expense: {max_expense} on Day {max_pos[0]+1}, Category: {categories[max_pos[1]]}")
print(f"Lowest Expense: {min_expense} on Day {min_pos[0]+1}, Category: {categories[min_pos[1]]}")
```

**Budget Exceed Alert**

```python
budget = np.array([2000, 800, 1000, 1200])
exceed_mask = total_per_category > budget
exceed_categories = categories[exceed_mask]
```

```python
print("Categories exceeding budget:", exceed_categories)
```

**Daily Total & Average**

```python
daily_total = np.sum(expenses, axis=1)
daily_avg = np.mean(expenses, axis=1)
```

```python
for day, total, avg in zip(days, daily_total, daily_avg):
    print(f"Day {day}: Total = {total}, Average = {avg:.2f}")
```

**Highest Spending Day**

```python
max_day_index = np.argmax(daily_total)
```

```python
print(f"Highest Spending Day: Day {days[max_day_index]} with {daily_total[max_day_index]} total expense")
```
