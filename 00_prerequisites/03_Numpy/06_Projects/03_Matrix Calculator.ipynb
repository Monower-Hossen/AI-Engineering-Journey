```python
import numpy as np
```

```python
# Sample Matrices
A = np.array([[2, 3], [1, 4]])
B = np.array([[5, 2], [3, 1]])
```

```python
print("Matrix A:\n", A)
```

```python
print("Matrix B:\n", B)
```

**Addition & Subtraction**

```python
add = A + B
sub = A - B
```

```python
print("A + B:\n", add)
```

```python
print("A - B:\n", sub)
```

**Matrix Multiplication**

```python
mul = np.dot(A, B)  # or A @ B
print("A x B:\n", mul)
```

**Transpose**

```python
A_T = A.T
print("Transpose of A:\n", A_T)
```

**Determinant**

```python
det_A = np.linalg.det(A)
print("Determinant of A:", det_A)
```

**Inverse (if exists)**

```python
if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("Inverse of A:\n", inv_A)
else:
    print("Matrix A is not invertible")
```

**Solve Linear Equations**

```python
b = np.array([5, 6])
x = np.linalg.solve(A, b)
print("Solution of Ax = b:", x)
```

**Eigenvalues & Eigenvectors**

```python
eig_vals, eig_vecs = np.linalg.eig(A)
print("Eigenvalues of A:", eig_vals)
```

```python
print("Eigenvectors of A:\n", eig_vecs)
```
